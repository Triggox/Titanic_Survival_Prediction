# Titanic Survival Prediction

Predicting passenger survival on the Titanic using Logistic Regression.

## 1. Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

sns.set_style('whitegrid')
%matplotlib inline
```

## 2. Load & Explore Data

```python
data = pd.read_csv('train.csv')
df = pd.DataFrame(data)
df.info()
```

### Survival Count Visualization

```python
survival_counts = df['Survived'].value_counts()

fig, ax = plt.subplots(figsize=(6, 5))
colors = ['#e74c3c', '#2ecc71']
bars = ax.bar(['Did Not Survive', 'Survived'], survival_counts.values, color=colors, edgecolor='black', linewidth=0.8)

for bar, count in zip(bars, survival_counts.values):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 5, str(count),
            ha='center', va='bottom', fontsize=14, fontweight='bold')

ax.set_title('Titanic Survival Count', fontsize=16, fontweight='bold', pad=15)
ax.set_ylabel('Number of Passengers', fontsize=12)
ax.set_ylim(0, max(survival_counts.values) * 1.15)
ax.tick_params(axis='both', labelsize=11)
plt.tight_layout()
plt.show()
```

## 3. Data Cleaning

```python
df.isnull().sum()
```

```python
df['Age'].isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Embarked'].isnull().sum()
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df = df.drop('Cabin', axis=1)
```

```python
df.info()
df.isnull().sum()
```

## 4. Feature Engineering

```python
df = df.drop('Name', axis=1)
df = df.drop('Ticket', axis=1)
df = df.drop('PassengerId', axis=1)
df.info()
```

```python
encoder = LabelEncoder()
df['Sex'] = encoder.fit_transform(df['Sex'])
```

```python
df = pd.get_dummies(df, columns=['Embarked'], dtype=int)
df.head()
```

```python
df.columns
```

## 5. Train-Test Split & Scaling

```python
x = df.drop('Survived', axis=1)
y = df['Survived']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
x_train[:5]
```

## 6. Model Training (Logistic Regression)

```python
model = LogisticRegression()
model.fit(x_train, y_train)
```

## 7. Evaluation

```python
predictions = model.predict(x_test)
predictions[::10]
```

```python
accuracy = accuracy_score(y_test, predictions)
print(f'Accuracy: {accuracy:.4f}')
```

```python
cm = confusion_matrix(y_test, predictions)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Not Survived', 'Survived'],
            yticklabels=['Not Survived', 'Survived'])
plt.title('Confusion Matrix', fontsize=14, fontweight='bold')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.tight_layout()
plt.show()
```

```python
print(classification_report(y_test, predictions))
```
